- Look it up against file `images-****.csv` to get the URI of IIIF manifest
- Potentially run a SPARQL query to get the related resource from the staging grp KG
- return the IIIF URL for the input image identifier 

```python
import pandas as pd
import yaml
```

```python
with open('rename-iiif/rename-alias-map.yml', 'r') as f:
    rename_alias_map = yaml.safe_load(f)
```

```python
def lookup_roboflow_image_id(image_id: str, images_df: pd.DataFrame):
    if '.jpg' in image_id:
        image_id = image_id.replace('.jpg', '')
    try:
        return f"{images_df.loc[image_id].iiif_url}/full/max/0/default.jpg"
    except:
        print(f'No IIIF manifest URI for image {image_id}')
        return None
```
